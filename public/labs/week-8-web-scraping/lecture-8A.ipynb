## Part 2: Responsible scraping

### Legal concerns

RadPad scraped the entirety of Craiglist, Craigslist sued RadPad, and they were [awarded $60 million](http://labusinessjournal.com/news/2017/apr/14/radpad-ordered-pay-605-million-judgment-craigslist/)

### Two types of legal issues

1. Copyright infringement
    - For example: pictures, rental listing text
2. Terms of Use violations
    - **Unauthorized**: Is scraping prohibited in the website’s terms of use?
    - **Intentional**: Was the person aware of the terms? Did they check an “I agree to these terms” box?
    - **Causes damage**: Did the scraping overload the website, blocking user access?

### Public-site scraping: the LinkedIn case

- Ruling from 2022 said that scraping data that is publicly accessible on the internet is not a violation of the Computer Fraud and Abuse Act
- Linkedin had sued a competitor for scraping publicly available information from user profiles

[More info on the case](https://techcrunch.com/2022/04/18/web-scraping-legal-court/)

### Problematic use cases

- The facial recognition startup Clearview AI scraped billions of photos from social media websites. They [recently settled](https://www.mediapost.com/publications/article/389525/facial-recognition-company-clearview-settles-priva.html) a class action lawsuit that alleged they violated privacy laws
- Web scraping at a massive scale has been a key ingredient in generating the training datasets for generative AI models like ChatGPT. Companies like OpenAI and Meta [have been sued](https://apnews.com/article/openai-lawsuit-authors-grisham-george-rr-martin-37f9073ab67ab25b7e6b2975b2a63bfe) by authors and other content creators for violating copyright laws.

### When is scraping more likely to be appropriate?

- .gov sites and, to a lesser degree, .edu sites
- Website owner has no business reason to protect the information
- Not prohibited in terms of use
- Limited number of requests
- Not too many requests all at once
- Done at night, when web traffic is low

### When is scraping less likely to be appropriate?

- search engines
- E-commerce sites (e.g. Zillow, Expedia, Amazon)
- Social media
- Prohibited in terms of use
- Large number of requests
- High frequency of requests

### HTML: HyperText Markup Language

- The language most websites are written in
- The browser knows how to read this language and renders the output for you
- HTML is what a web crawler will see

#### HTML tags

- There are a standard set of tags to define the different structural components of a webpage
- For example: 
    - `<h1>`, `<h2>` tags define headers
    - `<p>` tags define paragraphs
    - `<ol>` and `<ul>` are ordered and unordered lists

### CSS: Cascading Style Sheets

- A language for styling HTML pages
- CSS styles (also known as selectors) are applied to HTML tags based on their name, class, or ID.

#### Basic CSS selectors

- Class
    - e.g., `.red`
- ID
    - e.g., `#some-id`
- Tag
    - e.g., `p`, `li`, `div`

- **IDs:** unique identifiers
    - no two elements on a page will have the same ID.
- **Classes:** not unique
    - many elements will have the same class
    - a single element can have multiple classes

### The DOM: Document Object Model

- An interactive object tree created from the HTML tag hierarchy on a page
- Created by the browser

### The Network tab: finding data sources

The network tab tracks all of the content loaded by a website. Many modern websites that use data will "fetch" the data from an external (possibly nonpublic) data source. The Network tab will show you these sources, potentially allowing you to get the data without writing any scraping code at all.

## Part 5: Guided demo — Philadelphia city health centers

Let's scrape the **Locations and hours** table on the City of Philadelphia's public [City health centers page](https://www.phila.gov/services/mental-physical-health/city-health-centers/).

This replaces the unavailable Inquirer restaurant-inspection demo. The city page includes its table in the HTML response, so we can extract it with `requests` and BeautifulSoup without running JavaScript.

Open the live page in your browser. We will collect each center's name, address, services, phone information, and hours. You only need an internet connection and the `requests`, `beautifulsoup4`, and `pandas` packages; no local data file is needed. The number of locations and their details may change over time.

### Step 1: Request the HTML

Use one GET request with a timeout, then check for HTTP errors before parsing. A browser user agent helps avoid the HTTP 403 response this site can return to default automated clients. If access is denied, stop and inspect the response rather than repeatedly retrying.

````python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = "https://www.phila.gov/services/mental-physical-health/city-health-centers/"
r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=30)
r.raise_for_status()
````

````python
r.status_code
````

### Step 2: Parse the HTML with BeautifulSoup

Create the soup from the response content. BeautifulSoup lets us select elements using the same CSS selectors we see in the browser.

````python
soup = BeautifulSoup(r.content, "html.parser")
````

### Step 3: Inspect and select the table

On the live page, scroll to **Locations and hours**, right-click a center name, and choose **Inspect**. Move up the DOM tree to the enclosing `<table>`.

The table has the class `dcf-table`. Its column headings are in `<thead>`, while the location rows are in `<tbody>`.

- `select_one(selector)` returns the first matching element, or `None`.
- `select(selector)` returns a list of all matching elements.

See the [CSS selector notes](css-selectors.md) for more examples.

````python
table = soup.select_one("table.dcf-table")
if table is None:
    raise ValueError("Health center table not found. Inspect the live page for HTML changes.")

headers = [tag.get_text(" ", strip=True) for tag in table.select("thead th")]
headers
````

Check the headings before extracting records: an HTTP 200 response alone does not prove we downloaded the expected table.

````python
expected_headers = ["Name", "Address", "Services", "Phone", "Hours"]
if headers != expected_headers:
    raise ValueError(f"Unexpected table columns: {headers}. Inspect the page before continuing.")
````

#### Select the first location row

`tbody > tr:nth-of-type(1)` is a **CSS selector we write**, not text that appears as an HTML tag. You will not see that exact string in the Elements tab when you inspect the page.

Read the selector one piece at a time:

| Part | Meaning |
| --- | --- |
| `tbody` | The table body, containing the location rows. |
| `>` | Look for direct children of that element. |
| `tr` | A table row. |
| `:nth-of-type(1)` | Choose the first sibling of this tag type: here, the first row. |

**Follow the structure in Inspect:**

1. On the live city page, right-click **Health Center 1** and choose **Inspect**.
2. The highlighted element may be `<strong>` or its enclosing `<td>`.
3. Move upward through its parents: `<strong>` → `<td>` → `<tr>` → `<tbody>` → `<table>`. Expand the arrows beside collapsed elements as needed.
4. Notice that `<tr>` is a direct child of `<tbody>`. That relationship is what `tbody > tr` describes.

You should see a structure like this (other cells and rows are omitted):

```html
<table class="dcf-table ...">
  <tbody>
    <tr>
      <td data-label="Name">
        <strong>Health Center 1</strong>
      </td>
      <!-- Other cells in this row -->
    </tr>
    <!-- Other location rows -->
  </tbody>
</table>
```

A browser's **Copy → Copy selector** may generate a much longer selector containing positional parts such as `:nth-child(...)`. Our shorter selector is based on the structure we inspected; it does not need to match the browser's copied selector exactly.

In the code below, `table` already refers to the health centers table. We search **within that table**, rather than across the entire webpage.

````python
selector = "tbody > tr:nth-of-type(1)"
first_row = table.select_one(selector)
print(first_row.prettify())
````

### Step 4: Generalize the selector to all location rows

Remove `:nth-of-type(1)` to select every row. The `>` means a direct child. Starting from `table` keeps our search inside the table we already checked, and using `tbody` excludes the heading row.

**A simpler way to select the first row:** `select_one()` already returns only the first match, so this also works:

```python
first_row = table.select_one("tbody > tr")
```

Compare the three expressions:

| Expression | Result |
| --- | --- |
| `table.select_one("tbody > tr:nth-of-type(1)")` | The first location row, selected explicitly by position. |
| `table.select_one("tbody > tr")` | The first location row, because `select_one()` returns the first match. |
| `table.select("tbody > tr")` | A list of all location rows. |

Now use `select()` to collect all rows:

````python
rows = table.select("tbody > tr")
if not rows:
    raise ValueError("No health center rows found. Check the live HTML and selector.")
len(rows)
````

````python
row = rows[0]
````

### Step 5: Extract fields from one row

Each `<td>` has a `data-label` attribute identifying its column. Attribute selectors such as `td[data-label="Name"]` are easier to read than selecting a column by its position.

#### Field 1: Center name

````python
name_tag = row.select_one('td[data-label="Name"]')
name_tag.attrs
````

````python
name = name_tag.get_text(" ", strip=True)
name
````

#### Field 2: Address

Addresses contain `<br>` line breaks. `get_text(" ", strip=True)` inserts a space between text fragments and trims whitespace, so the street and city stay separated.

````python
address_tag = row.select_one('td[data-label="Address"]')
address = address_tag.get_text(" ", strip=True)
address
````

#### Fields 3–5: Services, phone information, and hours

````python
services = row.select_one('td[data-label="Services"]').get_text(" ", strip=True)
phone = row.select_one('td[data-label="Phone"]').get_text(" ", strip=True)
hours = row.select_one('td[data-label="Hours"]').get_text(" ", strip=True)
{"services": services, "phone": phone, "hours": hours}
````

#### Nested selectors and optional elements

Some centers list a separate dental phone number; others do not. Select the `<strong>` labels **inside the Phone cell**. Searching all `<strong>` tags in the row would also collect weekday names from Hours.

`select()` returns an empty list when there are no matching labels. `select_one()` would return `None`; check for it before accessing `.text`.

````python
phone_labels = row.select('td[data-label="Phone"] strong')
[label.get_text(" ", strip=True) for label in phone_labels]
````

### Step 6: Build a DataFrame

Loop over the same downloaded rows; we do not need another web request. Check required cells so a page redesign produces a useful error rather than silently recording missing values.

````python
data = []
fields = {
    "name": "Name",
    "address": "Address",
    "services": "Services",
    "phone": "Phone",
    "hours": "Hours",
}

for this_row in rows:
    record = {}
    for field, label in fields.items():
        tag = this_row.select_one(f'td[data-label="{label}"]')
        if tag is None:
            raise ValueError(f"Missing {label} cell. Inspect the live table structure.")
        record[field] = tag.get_text(" ", strip=True)
    data.append(record)

scraped_df = pd.DataFrame(data)
scraped_df
````

### Step 7: Prepare the addresses for spatial analysis

Extract the five-digit ZIP code from each address. Keep it as text because ZIP codes are identifiers. We can later geocode the street addresses to map the centers.

````python
scraped_df["zip_code"] = scraped_df["address"].str.extract(r"\b(\d{5})\b", expand=False)
scraped_df[["name", "address", "zip_code"]].sort_values(
    ["zip_code", "name"], ignore_index=True
)
````
