# Census spatial-query report

This small report reads the exports created by the DuckDB Spatial analysis. The query assigned each synthetic historical Census dot to one 2015 Pennsylvania district. These are sample-dot counts, not population estimates.

## Inspect the exported summary


```python
import pandas as pd
from IPython.display import display, Image

summary = pd.read_csv("assets/district_counts.csv", dtype={"district_id": "string"})
summary.head()
```

## Check the total

A boundary tie is assigned to the smallest district ID so one dot is counted once.


```python
print(f"{len(summary)} historical districts; {summary.sample_dots.sum():,} sample dots.")
```

## Map the query result

Colors represent sample-dot counts, not density or current district populations.


```python
display(Image(filename="assets/district_sample_map.png"))
```

## Limitations

Synthetic locations, systematic thinning, and historical category/boundary definitions limit interpretation. The map demonstrates the database-to-publication workflow. It is not suitable for demographic inference or evaluating current districts.

Sources: [historical Census dot archive](https://s3.amazonaws.com/datashader-data/census2010.parq.zip) and [2015 Census boundary files](https://www2.census.gov/geo/tiger/GENZ2015/shp/cb_2015_us_cd114_20m.zip).
